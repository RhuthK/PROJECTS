```python
import requests
import json
api_url="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json"

# A function that makes calls to API to get the total number of jobs for a particular technology 
def number_of_jobs(technology):
    num_jobs = 0
    response= requests.get(api_url)
    if response.ok:
        data = response.json()
        for i in data:
            if technology in i.get("Key Skills"):
                num_jobs += 1
            else:continue

    return num_jobs


# getting data for various technologies 
```

```python
from bs4 import BeautifulSoup
import requests

url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/datasets/Programming_Languages.html"
data = requests.get(url).text
soup = BeautifulSoup(data, 'html.parser')

# scrapign language list and saalry 
lang_list = []
salary = []

table = soup.find('table')

for row in table.find_all('tr'):
    col = row.find_all('td')
    lang_list.append((col[1]).text.strip())
    salary.append((col[3]).text.strip())
    

    
    
lang_list = lang_list[1:]
salary = salary[1:]

lang_list
```

```python

poplang_df = pd.DataFrame({"Languages": [a for a in lang_list], "salary": [x for x in salary]})
poplang_df=poplang_df.sort_values(by='salary', ascending = True)

poplang_df.to_csv('popular_languages.csv', header=False)
poplang_df
```

```python
poplang_df.salary = poplang_df.salary.replace({'\$':'', ',':''},regex=True).astype('int64')
poplang_df

# visualization
labels = list(poplang_df.Languages)
poplang_df.plot(kind='pie',
                      y='salary',
                      autopct = '%1.1f%%',
                labels=labels,
                    pctdistance=0.8,
                     cmap='viridis',
                     title="Top 10 Technologies with Highest Salaries",
                     ylabel='',
                      radius=1)
                     
                     
plt.legend(ncols=4, loc='lower left', bbox_to_anchor=(-0.1,-0.2))
plt.show()
```

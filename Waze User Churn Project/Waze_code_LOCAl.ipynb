<h2><b> Background and Business Problem</b></h2>
<p>Waze's navigation app serves millions of drivers worldwide, providing free navigation services through a collaborative community of users, map editors, and partners. However, like many digital platforms, Waze faces the challenge of user churn—users who uninstall the app or stop using it entirely. This creates significant business challenges that directly impact growth and sustainability.

The manual analysis of user behavior patterns was time-consuming and reactive, leading to several critical issues:

- <b>Reactive churn management:</b> Users were already lost before intervention strategies could be implemented
- <b>Inefficient resource allocation:</b> Marketing and retention efforts were spread broadly rather than targeted at high-risk users
- <b>Limited predictive capability:</b> Inability to identify early warning signs of potential churn
- <b>Missed retention opportunities: </b> High-value users at risk of churning went unidentified until it was too late

To address these challenges and support Waze's growth objectives, the data team aims to develop a predictive machine learning model that can identify users at high risk of churning before they actually leave the platform. This proactive approach will enable targeted retention strategies and improve overall user satisfaction.</p>

This dataset contains 82% retained users and 18% churned users. 

To understand a bit more about the each variables for churned and retained users we will calculate the median as mean is affected by outliers. 

```python
df.groupby(['label']).median(numeric_only=True)
```

The above gives a good snapshot of two groups, churned vs. retained: 

1) The users who churned had averaged 3 more drives ( from `drives` ) than the retained users but retained users use the app twice as many times as churned users (from `activity_days` ) in the same time period.
   
2) The median churned users drove ~200 more kilometres (from `driven_km_drives` ) and 2.5 hours more than the median retained user (from `duration_minutes_drives`).

   
3) Overall, it looks like churn users have more drives in fewer days and have driven more and for longer hours than the retained users. We can look more into this by taking a median kilometers per drive for both retained and churned users. 
   

```python
# How much average kilometres were driven per drive ?
df['km_per_drive'] = df.driven_km_drives / df.drives 
median_km_per_drive = df.groupby('label').agg({'km_per_drive': 'median'})
median_km_per_drive
```

The median retained user drove about one kilometer more on average per drive than the median churned user. 

```python
# How much average kilometers were driven per driving day ?   
df['km_per_driving_day'] = df.driven_km_drives / df.driving_days
median_km_per_driving_day = df.groupby('label').median(numeric_only=True)[['km_per_driving_day']]
median_km_per_driving_day
```

To understand the median number of drives per driving day for each group

```python
df['drives_per_driving_day'] = df.drives / df.driving_days
median_drives_per_driving_day = df.groupby('label').median(numeric_only=True)[['drives_per_driving_day']]
median_drives_per_driving_day
```

From above we can observe that: 

1) The median users who churned drove 698 kilometres each day than they drove last month which is **~240%** higher than the retained users. This is a lot of driving per drive day.
   
2) The median churned user had similarly disproportionate numbers of drives per drive day compared to the retained users, almost more than twice of retained users. 

Regardless of churned users or not the data represent users who are serious drivers! it would be safe to assume that this data does not represent typical drivers at large. Perhaps the data and particularly churned users data contains a high proportion of long-haul truckers. 

This information is worth mentioning to Waze as more data needs to be gathered on these serious drivers. It is possible that the reason for them driving so much is also the reason why Waze app does not meet specific requirement or set of needs, which differes from the need of a typical driver such as a commuter. This might also suggest a non random sampling where the sample does not represent typical drivers and only focus on long haul drivers.
